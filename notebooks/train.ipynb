```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Input, Embedding
from tensorflow.keras.callbacks import (EarlyStopping, ReduceLROnPlateau, ModelCheckpoint)
```

```python
# import training and testing data
import sys
sys.path.append("..")
from src.preprocess import prepare_training_data
```

```python
(X_train, X_test, y_train, y_test, tokenizer) = prepare_training_data()
```

```python
import joblib
```

```python
model = Sequential([
    Input(shape=(300,)),
    Embedding(input_dim=10000, output_dim=128, mask_zero=True),
    LSTM(128),
    Dense(1, activation='sigmoid')
])
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.optimizer
```

```python
model.loss
```

```python
early_stop = EarlyStopping(monitor='val_loss', patience = 3, restore_best_weights=True, verbose=1)
```

```python
model_check = ModelCheckpoint(filepath='models/best_model.keras', save_best_only = True, monitor='val_loss', verbose=1)
```

```python
reduce_lr = ReduceLROnPlateau(monitor='val_loss', patience=2, factor=0.5, min_lr=0.001,verbose=1)
```

```python
reduce_lr
```

```python
model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=10, batch_size=32, callbacks=[early_stop, model_check, reduce_lr], verbose=1)
```
